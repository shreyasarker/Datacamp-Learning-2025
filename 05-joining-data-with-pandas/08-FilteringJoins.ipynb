# Filtering Joins
 
Filtering joins focus on selecting rows, not combining columns. In this chapter, you learn how to perform filtering joins in pandas, even though pandas does not provide them directly.

So far, you have used mutating joins, which add columns from one table to another. In contrast, filtering joins keep or remove rows from one table based on whether matching rows exist in another table.

A semi join keeps only the rows from the left table that have a match in the right table. It returns only the left table’s columns and does not duplicate rows, even if there are multiple matches.

An anti join does the opposite. It returns rows from the left table that do not have a match in the right table. Like a semi join, it only includes columns from the left table.

Both semi joins and anti joins are useful when you want to filter data based on the presence or absence of related records in another table.
